### 3) إحصاء NumPy الجاهز مقابل الدوال اليدوية (يجب أن يتطابقوا)

```python
np.random.seed(42)
rnd = np.random.randint(1, 100, size=15)
print("random ints:", rnd)

# الدوال اليدوية (نفس فكرة تاسك 2)
def manual_mean(x):
    return sum(x) / len(x)

def manual_variance(x):
    m = manual_mean(x)
    return sum((v - m) ** 2 for v in x) / len(x)

def manual_std(x):
    return manual_variance(x) ** 0.5

rl = rnd.tolist()
print(f"mean : numpy={np.mean(rnd):.6f}  manual={manual_mean(rl):.6f}  match={np.isclose(np.mean(rnd), manual_mean(rl))}")
print(f"std  : numpy={np.std(rnd):.6f}  manual={manual_std(rl):.6f}  match={np.isclose(np.std(rnd), manual_std(rl))}")
print(f"max={rnd.max()}  min={rnd.min()}")
```

## الجزء 4 — تقرير استكشاف Titanic

### 1) نظرة عامة
داتاسيت تايتانيك يحتوي بيانات ركاب السفينة. أهم الأعمدة:
- **survived**: هل نجا الراكب؟ (0 = لا، 1 = نعم).
- **pclass**: فئة التذكرة (1 = أولى، 2 = ثانية، 3 = ثالثة) — مؤشر للطبقة الاجتماعية.
- **sex**: جنس الراكب.
- **age**: العمر بالسنوات.
- **fare**: سعر التذكرة المدفوع.
- **embarked / embark_town**: ميناء الصعود.
- **class / who / alive**: نسخ وصفية من الأعمدة السابقة.

```python
print("عدد الصفوف والأعمدة:", titanic.shape)
print("الأعمدة:", list(titanic.columns))
```

### 2) القيم المفقودة

```python
missing = titanic.isnull().sum()
print(missing[missing > 0].sort_values(ascending=False))
```

### 3) إحصاء وصفي (describe) وتفسير سريع

```python
desc = titanic.describe()
print(desc)
print(f"\nمتوسط سعر التذكرة: {titanic['fare'].mean():.2f}")
print(f"أكبر عمر: {titanic['age'].max():.0f}  |  أصغر عمر: {titanic['age'].min():.2f}")
```

### 4) أسئلة استكشافية (بالكود)

```python
rate = titanic["survived"].mean()
print(f"نسبة النجاة الإجمالية: {rate:.4f}  ({rate*100:.1f}%)")
```

```python
by_sex = titanic.groupby("sex")["survived"].mean()
print("نسبة النجاة حسب الجنس:")
print(by_sex)
```

```python
by_class = titanic.groupby("pclass")["survived"].mean()
print("نسبة النجاة حسب الفئة:")
print(by_class)
print(f"\nالفرق بين الفئة الأولى والثالثة: {by_class.loc[1]-by_class.loc[3]:.4f}")
```

```python
age_by_surv = titanic.groupby("survived")["age"].mean()
print("متوسط العمر: 0=لم ينجُ, 1=نجا")
print(age_by_surv)
```

### 5) الخلاصة — أهم 3 ملاحظات

1. **الجنس كان العامل الأقوى في النجاة:** نجت **74.2%** من الإناث مقابل **18.9%**
   فقط من الذكور، وهذا يعكس مبدأ "النساء والأطفال أولًا" في الإنقاذ.
2. **الطبقة الاجتماعية أثّرت بوضوح:** نسبة النجاة في الفئة الأولى **63%** تقريبًا،
   بينما في الفئة الثالثة **24%** فقط — أي أن ركاب الدرجة الأولى نجوا بنسبة تفوق
   ضعف ركاب الدرجة الثالثة.
3. **البيانات فيها نقص واضح وأعمار متقاربة:** عمود **deck** مفقود في 688 صفًا
   وعمود **age** في 177 صفًا، ونسبة النجاة الإجمالية كانت **38.4%** فقط. كما أن
   فرق العمر بين من نجا (28.3) ومن لم ينجُ (30.6) كان بسيطًا، فالعمر لم يكن عاملًا
   حاسمًا مثل الجنس والفئة.
